```python
# 1. Установка и импорт библиотек
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
import joblib

print("Библиотеки загружены!")
```

```python
# 2. Набор данных с отзывами
data = {
    'text': [
        "Отличная жвачка, мята очень долгая и приятная!",
        "Классический Orbit, всегда беру его после еды.",
        "Супер свежесть, зубы чистые, вкус держится долго.",
        "Самая лучшая мятная жвачка на рынке, рекомендую.",
        "Очень вкусная, освежает дыхание отлично.",
        "Ужасный вкус, через 2 минуты превращается в резину.",
        "Вкус пропал мгновенно, больше никогда не куплю.",
        "Слишком жесткая жвачка, челюсть устает жевать.",
        "Не понравилась, оставляет неприятное химическое послевкусие.",
        "Полное разочарование, старый засохший Orbit."
    ] * 20, # искусственно умножаем, чтобы было достаточно примеров для обучения
    'sentiment': [
        'Положительный', 'Положительный', 'Положительный', 'Положительный', 'Положительный',
        'Отрицательный', 'Отрицательный', 'Отрицательный', 'Отрицательный', 'Отрицательный'
    ] * 20
}

df = pd.DataFrame(data)
print(f"Всего отзывов: {len(df)}")
df.head()
```

```python
# 3. Обучение модели
X_train, X_test, y_train, y_test = train_test_split(
    df['text'], df['sentiment'], test_size=0.2, random_state=42
)

# Пайплайн: преобразование текста в числа (TF-IDF) + Логистическая регрессия
model = Pipeline([
    ('tfidf', TfidfVectorizer()),
    ('clf', LogisticRegression())
])

model.fit(X_train, y_train)

# 4. Проверка точности
y_pred = model.predict(X_test)
print(f"Точность модели (Accuracy): {accuracy_score(y_test, y_pred) * 100:.2f}%")
print("\nОтчет классификации:")
print(classification_report(y_test, y_pred))
```

```python
# 5. Проверка работы на новом тексте
sample_text = "Очень вкусно и свежо"
result = model.predict([sample_text])[0]
print(f"Тест: '{sample_text}' -> {result}")

# 6. Сохранение и скачивание модели
joblib.dump(model, 'model.pkl')

from google.colab import files
files.download('model.pkl')
print("Файл model.pkl скачивается на компьютер!")
```
