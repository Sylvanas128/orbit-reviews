# 🍬 Orbit Reviews — Обучение модели анализа отзывов

В этом блокноте происходит загрузка данных, предобработка текста с помощью TF-IDF, обучение модели Logistic Regression и её сохранение.

```python
# 1. Установка и импорт необходимых библиотек
import pandas as pd
import numpy as np
import joblib
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.metrics import classification_report, accuracy_score
print("Библиотеки успешно импортированы!")
```

```python
# 2. Создание демонстрационного датасета
data = {
    'text': [
        'Отличный товар, качество супер, рекомендую к покупке!',
        'Ужасный сервис, все сломалось на второй день, не покупайте.',
        'Очень доволен быстрой доставкой и упаковкой.',
        'Полное разочарование, зря потраченные деньги и время.',
        'Прекрасный магазин, вежливые менеджеры, товар качественный.',
        'Прислали брак, возвращать деньги отказываются, кошмар.'
    ],
    'sentiment': [1, 0, 1, 0, 1, 0]
}

df = pd.DataFrame(data)
X_train, X_test, y_train, y_test = train_test_split(df['text'], df['sentiment'], test_size=0.2, random_state=42)
print(f"Размер обучающей выборки: {len(X_train)}")
```

```python
# 3. Создание Пайплайна (TF-IDF + Логистическая регрессия) и обучение
model_pipeline = Pipeline([
    ('tfidf', TfidfVectorizer()),
    ('lr', LogisticRegression())
])

model_pipeline.fit(X_train, y_train)
print("Модель успешно обучена!")
```

```python
# 4. Оценка качества модели
predictions = model_pipeline.predict(X_test)
print(f"Accuracy: {accuracy_score(y_test, predictions)}")
print(classification_report(y_test, predictions))
```

```python
# 5. Сохранение обученной модели в файл model.pkl
joblib.dump(model_pipeline, 'model.pkl')
print("Модель успешно сохранена в файл model.pkl!")
```
